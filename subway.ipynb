## 2.1 지하철역의 위경도 파악

### googlegeocoding API 사용

```python
import pandas as pd
subway_station = pd.read_csv("C:/Users/chogu/Jupyter_bigdata/data/서울시 지하철 호선별 역별 시간대별 승하차 인원 정보.csv", encoding = "euc-kr")

#몇 호선인지와 지하철 역명 출력
subway_station = subway_station.loc[subway_station['사용월']==202505,:]
subway_station = subway_station[['호선명','지하철역']]

subway_station.head()
```

```python
subway_station.describe()
```

```python
# 뒤에 + "역"을 붙임
for i in subway_station.index:
    station=''.join(list(subway_station.loc[i,['지하철역']]))
    if station[-1] != '역':
        subway_station.loc[i,['지하철역']] = station+'역'

import googlemaps
my_key = "**********************"
maps = googlemaps.Client(key=my_key)
lat = []  #위도
lng = []  #경도

# 위치를 찾을 장소나 주소를 넣어줌
places = list(subway_station['지하철역'])

i=0
for place in places:   
    i = i + 1
    try:
        geo_location = maps.geocode(place)[0].get('geometry')
        lat .append(geo_location['location']['lat'])
        lng.append(geo_location['location']['lng'])
        

    except:
        lat.append('')
        lng.append('')
        print("%d번 인덱스 위치를 찾는데 실패했습니다."%(i))


# 데이터프레임 만들어 출력
dataframe = pd.DataFrame({'위도':lat, '경도':lng}, index=places)
print(dataframe)
```

```python
# 결측치(빈 문자열)인 행만 고르기
missing = dataframe[(dataframe['위도'] == '') | (dataframe['경도'] == '')]

# 결과 출력
print(f"결측치가 있는 역 개수: {len(missing)}\n")
print(missing)
```

```python
import pandas as pd
from folium.plugins import HeatMap
import folium

# 5개월 합·일평균 계산 (기존과 동일) —#
days = {202501:31, 202502:28, 202503:31, 202504:30, 202505:31}
total_days = sum(days.values())

period_df = df_subway[
    (df_subway['사용월'] >= 202501) &
    (df_subway['사용월'] <= 202505)
].copy()

# '승차' 컬럼 합 → 월별_승차합계
boarding_cols = [c for c in period_df.columns if '승차' in c]
period_df['월별_승차합계'] = period_df[boarding_cols].sum(axis=1)

# 역별 5개월 합 → 하루평균승차
station_tot = (
    period_df
    .groupby('지하철역', as_index=False)['월별_승차합계']
    .sum()
)
station_tot['하루평균승차'] = station_tot['월별_승차합계'] / total_days

# CSV에서 좌표 불러와 전처리 —#
coords = pd.read_csv("서울교통공사_역주소_위경도.csv", dtype=str)

# 역 이름 포맷 통일: 뒤에 '역' 붙이기 (뭐가 문제야!!!!!!!!!!!!!!!!!!!!!!!!!!!)
# station_tot의 역 이름과 서울교통공사_역주소_위경도.csv의 역 이름이 정확히 일치하지 않아서, 병합(merge)이 한 역에 대해서만 성공했고 그 역의 좌표가 시간대 개수(5개)만큼 중복 매칭
coords['지하철역'] = coords['지하철역'].str.strip()
station_tot['지하철역'] = station_tot['지하철역'].astype(str).apply(
    lambda x: x + '역' if not x.endswith('역') else x
)

# 중복된 역 이름은 하나만 남기기
coords = coords.drop_duplicates(subset='지하철역', keep='first')

# 위·경도 → 숫자형, NaN 제거
coords['위도'] = pd.to_numeric(coords['위도'], errors='coerce')
coords['경도'] = pd.to_numeric(coords['경도'], errors='coerce')
coords = coords.dropna(subset=['위도','경도'])

# 병합 & 중복 좌표 한번 더 제거 —#
heat_df = pd.merge(
    station_tot[['지하철역','하루평균승차']],
    coords[['지하철역','위도','경도']],
    on='지하철역',
    how='inner'
)

# 혹시 동일 좌표가 남아 있다면
heat_df = heat_df.drop_duplicates(subset=['위도','경도'])

print(f"HeatMap에 들어갈 포인트 개수: {len(heat_df)}")  # → 수백 개 나와야 정상

# HeatMap 그리기 —#
heat_data = heat_df[['위도','경도','하루평균승차']].values.tolist()

m = folium.Map(location=[37.541, 126.986], zoom_start=11)
HeatMap(
    heat_data,
    radius=15,
    blur=25,
    max_zoom=12
).add_to(m)
m.save('202501-05_역별_하루평균승차_heatmap_corrected.html')
```

```python
import pandas as pd
from folium.plugins import HeatMap
import folium

# 0) 기간·일수 설정
days = {202501:31, 202502:28, 202503:31, 202504:30, 202505:31}
total_days = sum(days.values())

# 1) 2025.01~05 데이터만 필터링
period_df = df_subway[
    (df_subway['사용월'] >= 202501) &
    (df_subway['사용월'] <= 202505)
].copy()

# 2) “하차” 관련 컬럼만 추출하여 월별 하차 합계 계산
alight_cols = [c for c in period_df.columns if '하차' in c]
period_df['월별_하차합계'] = period_df[alight_cols].sum(axis=1)

# 3) 역별 5개월 합 → 일평균 하차량
alight_tot = (
    period_df
    .groupby('지하철역', as_index=False)['월별_하차합계']
    .sum()
)
alight_tot['하루평균하차'] = alight_tot['월별_하차합계'] / total_days

# 4) CSV에서 위·경도 로드 & 전처리
coords = pd.read_csv("서울교통공사_역주소_위경도.csv", dtype=str)
coords['지하철역'] = coords['지하철역'].str.strip()
# station_tot 에도 '역'이 붙어있도록 통일
alight_tot['지하철역'] = alight_tot['지하철역'].apply(
    lambda x: x + '역' if not x.endswith('역') else x
)
# 중복 행 제거, 숫자형 변환, NaN 행 제거
coords = coords.drop_duplicates(subset='지하철역', keep='first')
coords['위도'] = pd.to_numeric(coords['위도'], errors='coerce')
coords['경도'] = pd.to_numeric(coords['경도'], errors='coerce')
coords = coords.dropna(subset=['위도','경도'])

# 5) 병합 & 중복 좌표 제거
heat_df_alight = pd.merge(
    alight_tot[['지하철역','하루평균하차']],
    coords[['지하철역','위도','경도']],
    on='지하철역',
    how='inner'
).drop_duplicates(subset=['위도','경도'])
print(f"HeatMap 포인트 수: {len(heat_df_alight)}")

# 6) HeatMap용 리스트 생성
heat_data = heat_df_alight[['위도','경도','하루평균하차']].values.tolist()

# 7) 지도 생성 및 레이어 추가
m = folium.Map(location=[37.541, 126.986], zoom_start=11)
HeatMap(
    heat_data,
    radius=15,    # 반경
    blur=25,      # 블러 강도
    max_zoom=12
).add_to(m)

# 8) 결과 저장
m.save('202501-05_역별_하루평균하차_heatmap.html')
```

```python
import pandas as pd

# 1) 데이터 불러오기
df = pd.read_csv(
    "C:/Users/chogu/Jupyter_bigdata/data/서울시 지하철 호선별 역별 시간대별 승하차 인원 정보.csv",
    encoding="euc-kr"
)

# 2) 컬럼 분류: “~승차”와 “~하차”
boarding_cols  = [c for c in df.columns if c.endswith("승차")]
alighting_cols = [c for c in df.columns if c.endswith("하차")]

# 3) 역별·시간대별 합계 계산
#    station_board: 호선명·지하철역을 인덱스로, 각 시간대별 승차 합
station_board  = (
    df
    .groupby(["호선명","지하철역"])[boarding_cols]
    .sum()
)

#    station_alight: 호선명·지하철역을 인덱스로, 각 시간대별 하차 합
station_alight = (
    df
    .groupby(["호선명","지하철역"])[alighting_cols]
    .sum()
)

# 4) 합친 뒤 인덱스 리셋
station_time = pd.concat([station_board, station_alight], axis=1) \
                   .reset_index()

# 5) (선택) 보기 편하게 시간대 컬럼 순서 정렬
#    예: '04-05승차', '04-05하차', '05-06승차', '05-06하차', ...
time_slots = sorted({col[:-2] for col in boarding_cols + alighting_cols})
ordered_cols = ["호선명","지하철역"] + \
               [f"{ts}승차" for ts in time_slots] + \
               [f"{ts}하차" for ts in time_slots]

station_time = station_time[ordered_cols]

# 결과 확인
print(station_time.head(10))
```
